#Import libraries

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
```

#Load dataset

```python
data = pd.read_excel('spam_dataset.xlsx')
print("Dataset loaded successfully!\n")
print(data.head())
```

#Encode labels (ham=0, spam=1)

```python
data['label_num'] = data['label'].map({'ham': 0, 'spam': 1})
```

#Split data

```python
X_train, X_test, y_train, y_test = train_test_split(
    data['message'], data['label_num'], test_size=0.3, random_state=42
)
```

#Text vectorization

```python
vectorizer = CountVectorizer(stop_words='english')
X_train_vec = vectorizer.fit_transform(X_train)
X_test_vec = vectorizer.transform(X_test)
```

#Train model

```python
model = MultinomialNB()
model.fit(X_train_vec, y_train)
```

#Evaluate

```python
y_pred = model.predict(X_test_vec)
print("\nModel Evaluation:\n")
print("Accuracy:", accuracy_score(y_test, y_pred))
print("\nClassification Report:\n", classification_report(y_test, y_pred))
```

#Confusion matrix

```python
print("\nConfusion Matrix:\n", confusion_matrix(y_test, y_pred))
```

#Predict custom messages

```python
def predict_spam(message):
    msg_vec = vectorizer.transform([message])
    prediction = model.predict(msg_vec)[0]
    return "Spam" if prediction == 1 else "Not Spam"
```

#Try a few examples

```python
examples = [
    "Win a brand new iPhone now!",
    "Can we have a meeting tomorrow morning?",
    "Get free coupons for your next purchase!"
]

for ex in examples:
    print(f"Message: {ex} -> Prediction: {predict_spam(ex)}")
```
